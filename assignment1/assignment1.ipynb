Can a pedestrian model handle two groups walking against each other
through a corridor? This is a classic verification test for evacuation

## Setup

- Two square rooms, each 10 m × 10 m.
- A corridor, 10 m long and 2 m wide, connects the middle of the right
  wall of room 1 with the middle of the left wall of room 2.
- Models: you compare two operational models,
  `jps.CollisionFreeSpeedModel` and `jps.SocialForceModel` (the model of
  the live example). Use the same setup for both.
- Group A: 100 agents in room 1, placed from the left wall with the
  highest density you can reach. Pre-evacuation time 0 s. Their target
  is room 2.
- Choose realistic desired speeds and say where your values come from.

**Measured quantity:** the time when the last agent of group A enters
room 2.

## 1. Geometry

Build the walkable area and plot it with axis labels in metres.

## 2. One run with group A only

Place group A, give it a journey to room 2, run the simulation and measure the time defined above. Watch the animation once.

## 3. Counter-flow

Add group B in room 2, walking at the same time towards room 1, with the
same characteristics as group A. Run the cases:

| Group B size | 0 | 10 | 50 | 100 |
|---|---|---|---|---|

Write a function that runs one case for a given model, group B size and
random seed, so you do not copy and paste code.

## 4. Repetitions and gridlock

- Repeat each case with several random seeds (positions and speeds).
- Stop every run at a maximum simulated time that you choose and justify.
- A run that reaches the maximum time is **gridlocked**. Count gridlocked
  runs per case. Do **not** put their capped times into the mean.
- If a run stops with an error, do not hide it: catch it, count it per
  case and find out what happened.

## 5. Plot and interpretation

- Plot the measured time against the size of group B for **both models**
  in one figure, with error bars over the completed runs. Mark gridlocked
  and failed runs clearly.
- In a few sentences: how does the time depend on the size of group B?
  Is what you see plausible for real people? If not, is it a property of
  the model or of your setup? How would you find out?
- Compare the two models: where do they agree, where do they differ, and
  which result would you trust for a safety decision? Why?
